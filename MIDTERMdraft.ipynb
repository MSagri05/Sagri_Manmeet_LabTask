## ques 2 part b

```python
# define the confusion matrix values from the question

tp = 70   # actually qualified and predicted qualified
fp = 20   # underqualified but predicted qualified
tn = 880  # underqualified and correctly predicted underqualified
fn = 30   # qualified but predicted underqualified

# calculate the overall accuracy of the model
# this tells us how many predictions were correct overall

accuracy = (tp + tn) / (tp + tn + fp + fn)

# calculate precision
# this tells us out of everyone predicted as qualified,
# how many were actually qualified

precision = tp / (tp + fp)

# calculate recall
# this tells us out of all the actually qualified applicants,
# how many the model successfully found

recall = tp / (tp + fn)

# print each metric as a percentage with 2 decimal places

print(f"accuracy : {accuracy:.2%}")
print(f"precision: {precision:.2%}")
print(f"recall   : {recall:.2%}")
```

```python
# import libraries
import pandas as pd
import matplotlib.pyplot as plt

# load the dataset
df = pd.read_csv("workstudy_pilot.csv")

# look at the first few rows
df.head()
```

```python
# count missing values in each column
df.isnull().sum()
```

```python
# see all unique values in form_type
df["form_type"].value_counts()
```

```python
# show application scores outside the valid range (0-100)

df[(df["application_score"] < 0) | (df["application_score"] > 100)]
```

```python
# standardize all form_type values

df["form_type"] = df["form_type"].replace({
    "Manual":"Manual_Entry",
    "manual_entry":"Manual_Entry",
    "LINKEDIN":"LinkedIn_Import",
    "linkedin_import":"LinkedIn_Import"
})

df["form_type"].value_counts()
```

```python
# histogram of application scores

plt.hist(df["application_score"], bins=10)
plt.title("application score distribution")
plt.xlabel("application score")
plt.ylabel("count")
plt.show()
```

```python
# average score for each group

df.groupby("form_type")["application_score"].mean()
```

```python
# load the dataset
df = pd.read_csv("phigma_render_logs.csv")

# first i want to see what the data looks like
print("first few rows:")
display(df.head())

# now i check the size of the dataset
print("shape of dataset:")
print(df.shape)

# now i check column types and missing values
print("dataset info:")
df.info()

# now i check summary stats to see min, max, mean etc
print("summary statistics:")
display(df.describe())

# check missing values in each column
print("missing values:")
display(df.isnull().sum())

# check duplicate sessions
print("duplicate session ids:")
print(df["session_id"].duplicated().sum())

# check if total_vector_nodes is stored as a number or text
print("data types:")
display(df.dtypes)

# try converting columns to numeric just in case some numbers are stored as text
df["total_vector_nodes"] = pd.to_numeric(df["total_vector_nodes"], errors="coerce")
df["imported_png_count"] = pd.to_numeric(df["imported_png_count"], errors="coerce")
df["frames_per_second"] = pd.to_numeric(df["frames_per_second"], errors="coerce")

# check invalid values based on what the question says
# fps should be between 4.1 and 60.0
# vector nodes and png count should not be negative
invalid_rows = df[
    (df["frames_per_second"] < 4.1) |
    (df["frames_per_second"] > 60) |
    (df["total_vector_nodes"] < 0) |
    (df["imported_png_count"] < 0)
]

print("invalid rows:")
display(invalid_rows)

# quick scatter plot to see relationship between vector nodes and fps
plt.scatter(df["total_vector_nodes"], df["frames_per_second"], alpha=0.6)
plt.title("total vector nodes vs frames per second")
plt.xlabel("total vector nodes")
plt.ylabel("frames per second")
plt.show()

# another scatter plot with png count because that can also affect performance
plt.scatter(df["imported_png_count"], df["frames_per_second"], alpha=0.6)
plt.title("imported png count vs frames per second")
plt.xlabel("imported png count")
plt.ylabel("frames per second")
plt.show()
```
